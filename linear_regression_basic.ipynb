```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from google.colab import files
uploaded= files.upload()
data=pd.read_csv('Salary_Data.csv')
```

```python
print('first five records')
print(data.head())
```

```python
x=data[['YearsExperience']]
y=data['Salary']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
model=LinearRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print("\n slope coefficient:",model.coef_)
print("\n intercept:",model.intercept_)
mse=mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
```

```python
experience = pd.DataFrame({'YearsExperience': [7]})
predict_salary = model.predict(experience)
print("\n Predicted salary for 7 years: {:.2f}".format(predict_salary[0]))
print("\n Mean squared error:", mse)
print("R2 score:", r2)
plt.scatter(x_test, y_test, color='blue', label='Test Data')
plt.plot(x_test, y_pred, color='red', linewidth=2, label='Regression Line')
plt.xlabel('YearsExperience')
plt.ylabel('Salary')
plt.title('Salary vs YearsExperience (Test Set)')
plt.legend()
plt.show()
plt.scatter(x_train, y_train, color='green', label='Training Data')
plt.plot(x_train, model.predict(x_train), color='red', linewidth=2, label='Regression Line')
plt.title('Salary vs YearsExperience (Training Set)')
plt.xlabel('YearsExperience')
plt.ylabel('Salary')
plt.legend()
plt.show()
```
